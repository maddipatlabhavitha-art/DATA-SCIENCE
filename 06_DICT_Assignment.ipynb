```python
# Question 43: Create a dictionary with nested lists: {'matrix': [[1, 2], [3, 4]], 'vector': [5, 6, 7]}
data = {
    'matrix': [[1, 2], [3, 4]],
    'vector': [5, 6, 7]
}

print(data)
```

```python
# Question 44: Find sum of each nested list
data = {
    'matrix': [[1, 2], [3, 4]],
    'vector': [5, 6, 7]
}

sums = {}

for key, value in data.items():
    # If nested list, sum each sublist
    if all(isinstance(i, list) for i in value):
        sums[key] = [sum(sublist) for sublist in value]
    else:
        sums[key] = sum(value)

print(sums)

```

```python
# Question 49: Create a dictionary with nested sets: {'groups': {{1, 2, 3}, {4, 5, 6}}, 'categories': {{'a', 'b'}, {'c', 'd'}}}
data = {
    'groups': {frozenset({1, 2, 3}), frozenset({4, 5, 6})},
    'categories': {frozenset({'a', 'b'}), frozenset({'c', 'd'})}
}

print(data)
```

```python
# Question 50: Find union of all nested sets
data = {
    'groups': {frozenset({1, 2, 3}), frozenset({4, 5, 6})},
    'categories': {frozenset({'a', 'b'}), frozenset({'c', 'd'})}
}

# Union of all nested sets
union_groups = set()
for s in data['groups']:
    union_groups |= s  # union with each frozenset
print("Union of groups:", union_groups)

union_categories = set()
for s in data['categories']:
    union_categories |= s
print("Union of categories:", union_categories)
```
