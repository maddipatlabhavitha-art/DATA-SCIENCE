```python
"""Pattern Printing - Right Triangle (Easy)
Print the following pattern for n rows:"""


n = int(input("Enter number of rows: "))
for i in range(1, n + 1):
    print('*' * i)

```

```python
"""Pattern Printing - Inverted Right Triangle (Easy)
Print the following pattern for n rows:"""

n = int(input("Enter number of rows: "))
for i in range(n, 0, -1):
    print('*' * i)
```

```python
"""Pattern Printing - Number Pattern (Medium)
Print the following number pattern for n rows:"""
n = int(input("Enter number of rows: "))
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end='')
    print()

```
