```python
# 1. Print numbers from 1 to 5
for i in range(1,5):
  print(i)
```

```python
# 2. Print each letter in a string
for i in "hello":
  print(i)
```

```python
# 3. Sum of numbers in a list
numbers=[2,3,4,5]
sum=0
for num in numbers:
  sum=sum+num
print(sum)
```

```python
# 4. Print even numbers from 1 to 10
for i in range(1,11):
  if i%2==0:
    print(i)
```

```python
# 5. Print multiplication table of 3
n=3
for i in range(1,11):
  print(n*i)
```

```python
# 6. Iterate over list with index
fruits=['APPLE','BANANA','ORANGE','CHERRY']
for i in range(len(fruits)):
  print(i,fruits[i])
```

```python
# 7. Print squares of numbers
for i in range(5):
    print(i**2)
```

```python
# 8. Loop over dictionary
student={"name": "alice", "age": 20}
for key in student:
  print(key,student[key])
```

```python
# 9. Nested loop for pairs
for i in range(1,4):
  for j in range(1,3):
    print(i,j)
```

```python
# 10. Loop over a set
colors={"red","green","blue"}
for color in colors:
  print(color)
```

```python
# 11. Loop with break
for i in range(1,5):
  if i==3:
    break
print(i)
```

```python
# 12. Loop with continue
for i in range(0,5):
  if i ==2:
    continue
print(i)
```

```python
# 13. Reverse a string
str="bhavitha"
a=str[::-1]
print(a)
```

```python
# 14. Loop through two lists
names=["bhavi","jaya","pinky"]
scores=[50,60,70]
for name,score in zip(names,scores):
  print(name,score)
```
