```python
"""Reverse a Number (Medium)
Write a program that reverses the digits of a given number using a while loop."""


num = int(input("Enter an integer: "))


is_negative = num < 0
num_abs = abs(num)

rev = 0


while num_abs > 0:
    digit = num_abs % 10
    rev = rev * 10 + digit
    num_abs //= 10

if is_negative:
    rev = -rev

print(f"Reversed number: {rev}")
```

```python
"""Problem 3.2: Sum of Digits (Easy)
Write a program that calculates the sum of all digits in a given number."""

num = int(input("Enter an integer: "))


num_abs = abs(num)


sum_digits = 0


while num_abs > 0:
    digit = num_abs % 10
    sum_digits += digit
    num_abs //= 10

print(f"Sum of digits of {num} is {sum_digits}")
```

```python
"""Problem 3.4: Palindrome Checker (Medium)
Write a program to check if a given number is a palindrome or not."""

num = int(input("Enter an integer: "))

if num < 0:
    print(f"{num} is not a palindrome.")
else:

    original = num
    reversed_num = 0


    while num > 0:
        digit = num % 10
        reversed_num = reversed_num * 10 + digit
        num //= 10


    if original == reversed_num:
        print(f"{original} is a palindrome.")
    else:
        print(f"{original} is not a palindrome.")
```
